```python
#generate positive list of numberts from a given list of integers
a=list(map(int,input("Enter numbers separated by space:").split()))
positive=[n for n in a if n>0]
print("Positive nmbers:",positive)
```

```python
#square of N numbers
n=int(input("Enter N:"))
a=[int(input("Enter number:"))for i in range(n)]
square=[x*x for x in a]
print("Squares:",square)

```

```python
#form a list of vowels selected from a given word
word=input("Enter a word:")
vowels=[ch for ch in word if ch in "aeiouAEIOU"]
print("Vowels:",vowels)

```

```python
#list ordinal value of each element of a word
word=input("Enter a word :")
values=[ord(ch) for ch in word]
print("Ordinal values:",values)
```
