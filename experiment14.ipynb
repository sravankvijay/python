```python
%%writefile graphics/rectangle.py
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
%%writefile graphics/circle.py
import math
def area(r):
    return round(math.pi*r*r,2)
def perimeter(r):
    return round(2*math.pi*r,2)

```

```python
%%writefile graphics/graphics3D/cuboid.py
def area(l,b,h):
    return 2*((l*b)+(l*h)+(b*h))
def perimeter(l,b,h):
    return 4*(l+b+h)

```

```python
%%writefile graphics/graphics3D/sphere.py
import math
def area(r):
    return 4*math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
from graphics import rectangle,circle
from graphics.graphics3D import cuboid,sphere
while True:
    print("------GRAPHICS CALCULATOR------")
    print("1.Rectangle")
    print("2.Circle")
    print("3.Cuboid")
    print("4.Sphere")
    print("5.Exit")
    choice=int(input("Enter your choice:"))
    if choice==1:
        l=float(input("Enter the length of Rectangle:"))
        b=float(input("Enter the breadth of Rectangle:"))
        print("Area of Rectangle : ",rectangle.area(l,b))
        print("Perimeter of Rectangle : ",rectangle.perimeter(l,b))
        print()
    elif choice==2:
        r=float(input("Enter the radius of Circle:"))
        print("Area of Circle : ",round(circle.area(r),2))
        print("Perimeter of circle : ",round(circle.perimeter(r),2))
        print()
    elif choice==3:
        l=float(input("Enter the length of Cuboid:"))
        b=float(input("Enter the breadth of Cuboid:"))
        h=float(input("Enter the height of Cuboid:"))
        print("Total Surface Area of Cuboid : ",cuboid.area(l,b,h))
        print("Perimeter of Cuboid : ",cuboid.perimeter(l,b,h))
        print()
    elif choice==4:
        r=float(input("Enter the radius of Sphere:"))
        print("Surface Area : ",round(sphere.area(r),2))
        print("Perimeter : ",round(sphere.perimeter(r),2))
        print()
    elif choice==5:
        print("Exiting program...")
        break
        
```
