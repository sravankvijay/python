```python
#finding the length of the logest word in a list
def longest_word(words):
    longest=max(words,key=len)
    return len(longest)
words=input("Enter words seperated by spaces:").split()
print("Length of longest word:",longest_word(words))
```

```python
#printing a star pattern using nested loops
def pattern():
    for i in range(1,6):
        for j in range(i):
            print("*",end="")
        print()
    for i in range(4,0,-1):
        for j in range(i):
            print("*",end="")
        print()
pattern()
```

```python
#Generating all factors of a number
def factors(n):
    for i in range(1,n+1):
        if n%i==0:
            print(i,end=" ")
n=int(input("Enter a number:"))
print("Factors are:")
factors(n)
```

```python
#Calculating the areas of Square,Rectangle and Traingle using Lambda Functions
def area():
    square=lambda a:a*a
    rectangle=lambda l,b:l*b
    triangle=lambda b,h:0.5*b*h
    print("Area of square:",square(4))
    print("Area of rectangle:",rectangle(5,3))
    print("Area of triangle :",triangle(6,4))
area()
```
