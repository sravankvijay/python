```python
%%writefile Rectangle.py

#Module to calculate the area and perimeter of a rectangle

def area(l,b):
    return   l*b
def perimeter(l,b):
    return  2*(l+b)
```

```python
%%writefile Circle.py
#Module to calculate area and perimeter of a circle
import math
def area(r):
    return round(math.pi*r*r,2)
def perimeter(r):
    return round(2*math.pi*r,2)
```

```python


#program to calculate area and perimeter of rectangle and circle using imported modules
import Rectangle
import Circle
while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")
    ch=int(input("Enter your choice:"))
    if ch==1:
        l=int(input("Enter the length of Rectangle:"))
        b=int(input("Enter the breadth of Rectangle:"))
        print("Area of Rectangle : ",Rectangle.area(l,b))
        print("Perimeter of Rectangle : ", Rectangle.perimeter(l,b))
        print()
    elif ch==2:
        r=int(input("Enter the radius of Circle:"))
        print("Area of Circle : ", round(Circle.area(r),2))
        print("Perimeter of circle : ", round(Circle.perimeter(r),2))
        print()
    elif ch==3:
        print("Exiting program...")
        break
        

```
